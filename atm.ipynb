```python
class BankATM:
    """
    Class representing a Bank ATM system with secure PIN validation,
    balance tracking, deposits, withdrawals, and transaction logging.
    """
    def __init__(self, account_holder, balance=1000.0, pin=1234):
        # Initialize user details and account state
        self.account_holder = account_holder
        self.balance = balance
        self.pin = pin
        self.transaction_history = []  # Stores log of all activities

    def verify_pin(self, input_pin):
        """Validates if the provided PIN matches the account PIN."""
        return input_pin == self.pin

    def check_balance(self):
        """Displays the current account balance."""
        print(f"\n----------------------------------------")
        print(f" Account Holder: {self.account_holder}")
        print(f" Current Balance: ${self.balance:,.2f}")
        print(f"----------------------------------------")

    def deposit(self, amount):
        """Deposits a specified amount into the account."""
        if amount > 0:
            self.balance += amount
            log_message = f"Deposited: ${amount:,.2f}"
            self.transaction_history.append(log_message)
            print(f"\n[SUCCESS] Deposited ${amount:,.2f}.")
            print(f"New Balance: ${self.balance:,.2f}")
        else:
            print("\n[ERROR] Deposit amount must be greater than $0.")

    def withdraw(self, amount):
        """Withdraws a specified amount if sufficient funds exist."""
        if amount <= 0:
            print("\n[ERROR] Withdrawal amount must be greater than $0.")
        elif amount > self.balance:
            print(f"\n[DENIED] Insufficient funds! Current balance: ${self.balance:,.2f}")
        else:
            self.balance -= amount
            log_message = f"Withdrew: ${amount:,.2f}"
            self.transaction_history.append(log_message)
            print(f"\n[SUCCESS] Withdrew ${amount:,.2f}.")
            print(f"New Balance: ${self.balance:,.2f}")

    def print_statement(self):
        """Displays all recorded transactions for the session."""
        print(f"\n----------------------------------------")
        print(f" TRANSACTION HISTORY")
        print(f"----------------------------------------")
        if not self.transaction_history:
            print(" No transactions performed yet.")
        else:
            for index, item in enumerate(self.transaction_history, start=1):
                print(f" {index}. {item}")
        print(f"----------------------------------------")


def run_atm_simulation():
    """Main execution control loop for the interactive ATM program."""
    
    # Create an instance of the BankATM class (Default PIN is set to 1234)
    user_account = BankATM(account_holder="User", balance=2500.00, pin=1234)

    print("========================================")
    print("      WELCOME TO THE VIRTUAL ATM        ")
    print("========================================")

    # --- Step 1: Authentication Phase ---
    attempts = 3
    authenticated = False

    while attempts > 0:
        try:
            entered_pin = int(input("\nEnter your 4-digit PIN (Default: 1234): "))
            if user_account.verify_pin(entered_pin):
                authenticated = True
                print("\n[AUTHENTICATED] PIN verified successfully.")
                break
            else:
                attempts -= 1
                print(f"[DENIED] Incorrect PIN. Attempts remaining: {attempts}")
        except ValueError:
            print("[ERROR] Please enter digits only.")

    if not authenticated:
        print("\n[LOCKED] 3 incorrect attempts. Card locked for security.")
        return

    # --- Step 2: Main Menu Loop ---
    while True:
        print("\n========================================")
        print("              MAIN MENU                 ")
        print("========================================")
        print("1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. View Statement / History")
        print("5. Exit & Remove Card")
        print("========================================")

        choice = input("Select an option (1-5): ").strip()

        if choice == "1":
            user_account.check_balance()

        elif choice == "2":
            try:
                deposit_amt = float(input("Enter amount to deposit: $"))
                user_account.deposit(deposit_amt)
            except ValueError:
                print("\n[ERROR] Invalid input. Enter numbers only.")

        elif choice == "3":
            try:
                withdraw_amt = float(input("Enter amount to withdraw: $"))
                user_account.withdraw(withdraw_amt)
            except ValueError:
                print("\n[ERROR] Invalid input. Enter numbers only.")

        elif choice == "4":
            user_account.print_statement()

        elif choice == "5":
            print("\nThank you for using the Virtual ATM service. Goodbye!")
            break

        else:
            print("\n[ERROR] Invalid selection. Please enter a number from 1 to 5.")


# Run the application
run_atm_simulation()
```
